# Modelling AFL Injury Risk Over a Full Season

# Abstract
Australian Rules Football is a contact sport in Australia and its professional league is called the AFL. The injury risk of a player is described as the probability of a player getting injured. The paper investigated whether players with a higher initial injury risk will have a more substantial reduction of injury risk compared to players with a lower initial injury risk. Injury risk was calculated using three main factors: recovery time, training load and match intensity. A stochastic, discrete and dynamic mathematical model was created and was implemented via a series of python functions. The model aimed to simulate and calculate injury risks across a 24-round season. There were two distinct groups of python functions, python functions that contained code where coaches could adjust the physical efforts of players based on their previous injury risk and functions that didn't contain such intervention. To fully test the model, it was run with various parameters and for each run, the functions were implemented 1000 times to account for the stochastic nature of the model. When testing with both types of functions, it was discovered that players who are injury prone and have a higher starting injury risk show a greater improvement of injury risk over a full season compared to players with a lower starting injury risk. This was likely due to the fact that players who have higher injury risks have more opportunities to rehabilitate their bodies compared to players with lower levels of injury risk. To extend this investigation further, the model can account for players who are not frequently selected as semi-professional and amateur players. 

# Introduction

The research will aim to investigate whether AFL players with a higher initial injury risk will show a greater reduction of injury risk compared to players who have a lower initial injury risk. While previous literature(Hrysomallis), has studied the most common injuries that occur amongst AFL players and how training load can impact a player's rate of injury(Whalan et al.), no research has specifically explored whether players with a higher initial injury risk show greater signs of improvement in their injuries, compared to players with a lower initial injury risk. Therefore, this paper aims to address this gap of knowledge. The model estimates the injury risk of AFL players over a 24-round season. It will investigate how key factors such as training load, match intensity and recovery time contribute to injury risk. Coaches adjust their training load of players depending on the match intensity and prior injury risk of players, while players are responsible for managing their own recovery. During an AFL match, players can either get injured through physical contact, or by accident (e.g. a hamstring injury that occurs unexpectedly). The purpose of the model is to investigate how injury risk of players changes dynamically over various factors during a full AFL season. It is predicted that players who are injury prone and thus have a high injury risk will end up having a higher improvement of injury risk over a season compared to players with lower levels of injury risk because they will have more opportunities to participate in rest and rehabilitation activites and won't be exposed to many contact opportunities, compared to players with a lower starting injury risk.


# Background
Australian Rules Football is a contact sport played in Australia. It is played between two teams of 18 players. The professional league is called the Australian Football League (AFL) and consists of players who both play and train full-time. The injury risk of players can be described as the likelihood of a player getting injured. It is determined by three main variables: training load, match intensity, and recovery time.  

Training load is measured in the number of extreme efforts per hour. Extreme efforts include rapid sprints, heavy contact drills and intense physical exercise(Mooney et al.). This is set directly by the coaches depending on previous team performance and the injury level of a player. As established earlier, depending on how high the risk of injury is for a player, coaches may modify training by reducing the number of high-intensity drills. This ensures the player’s rate of injury doesn’t deteriorate. As matches vary each week depending on the opponent, coaches adjust the training of players so that their bodies can with-stand a full season. Training load is dependent upon the previous values of match intensity, recovery and injury risk. This captures how coaches monitor the changes of the body that a player endures in a season and adjusts their plan accordingly.  

Match intensity is measured by the number of physical contests per hour. Physical contests can be broken down into tackles, bumps and collisions. These lead to higher chances of injury because high-impact tackles can lead to injuries such as soft-tissue damage, hamstring strains and calf-tears(“Australian Football AFL | Maitland Myotherapy”). Additionally, the 360-degree nature of the game increases the likelihood of concussions. Match intensity is determined by the previous values of injury risk and training load. The addition of injury risks reflects tactical decisions by coaches to manage the minutes of players who have a previously established high risk of injury. This was demonstrated in the 2025 AFL Grand Final, when Lachie Neale, a Brisbane Lions player who had been suffering from a calf-injury, was only restricted to playing in the third and fourth quarter only, to maximise his impact on the field. The previous training load can affect match intensity because if a training load is too strenuous for a player, this may subsequently affect the match intensity. This is further multiplied by the average speed of the player, because an increase in speed is attributed to an increase in the number of collisions which leads to an increase in injury risk. 

Recovery time is measured in hours. It is primarily dependent on sleep and nutrition. Sleep is measured by the number of hours a player sleeps. Nutrition level is measured from a scale from 0 to 10. 0 represents the fact that a player disregards their health and eats junk food all of the time, while 10 represents a player who sticks to a healthy, regimented diet set by their club dieticians. This is shown to be affected also by injury risk as players take longer time to recover if they are injured more. If a match is heavily congested and players are involved in many high-effort sprints throughout the game, they may need more time to recover subsequently. 

# Method 

**Definition of Key Mathematical Terms**

Let $r$ represent the rth round (e.g. round 1, round 2)

Let $i_r$ be the injury risk for the rth round. 

Let $m_r$ be the match intensity for the rth round

Let $m_{\mathrm{final}(r)}$ be the match intensity calculation after adding additional components

Let $t_r$ be the training load for the rth round

Let $t_{\mathrm{final}(r)}$ be the training load calculation after adding additional components

Let $R_{\mathrm{final}(r)}$ be the recovery time calculation after adding additional components

Let $\mathrm{Uniform} (xy, ab)$ represents random factors that is outside the player's control that causes an injury. xy and ab represent the interval values of the distribution

Let $s$ represent sleep

Let $n$ represent nutrition level

Let $m_{collisions}$ represent the number of collisions during a match

Let $m_{bumps}$ represent the number of bumps during a match

Let $m_{tackles}$ represent the number of tackles during a match

Let $m_{speed}$ represent the average speed during a match

Let $t_{total}$ represent the number of physical exertions during training

Let $distance$ be the total distance a player a traversed during a game

Let $T$ be duration of activity in hours

Let $o, a, b, c, h$ be constants of equations belonging to the injury risk equation

Let $d, e, f$ be constants of equations belonging to the match intensity equation

Let $s, x, y$ be constants of equations belonging to the recovery time equation

Let $q, i, j$ be constants of equations belonging to the training load equation

**Mathematical Representation of Model**

Injury risk can be modelled as:

$$ i_r = o \cdot \frac{1}{1 + e^{-(a \cdot t_{final(r)} + b \cdot m_{final(r)} -  c \cdot R_{final(r)} + \mathrm{Uniform}(xy, ab))}} + h \cdot i_{r-1} + \mathrm{Uniform} (xy, ab) $$

Speed can be calculated as

$$ m_{speed} = distance / T $$

Match intensity can be modelled as 

$$ m_{total} = m_{\text{tackles}} + m_{\text{bumps}} + m_{\text{collisions}}  $$

$$ m_r = m_{total} / T  $$

$$ m_{final(r)} = m_r \cdot \frac{m_{\text{speed}} - 14}{30} + d \cdot e^{-i_{r-1}} + f \cdot {t_{final(r - 1)}}$$

Recovery time can be modelled as 

$$ R_{final(r)} = s \cdot \frac{n}{10} + x \cdot  i_{r-1} + y \cdot {m_{final(r - 1)}}$$

Training load can be modelled as

$$ t_r = t_{total} / T$$

$$ t_{final(r)} = t_r + q \cdot e^{-i_{r-1}} + i \cdot {m_{final(r - 1)}} + j \cdot {R_{final(r - 1)}} $$


**Code Representation of Model**

```python
import random
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd 

def calculate_match_intensity(injury_risk, training_load):
    "Calculates match intensity for one round"
    tackles = random.randint(0, 8)
    bumps = random.randint(0, 5)
    collisions = random.randint(0, 30)
    # If a coach manages a player with serious injury
    if injury_risk > 0.5:
        distance = random.randint(11, 16)
        time = random.uniform(1, 1.5)
    else:
        distance = random.randint(11, 19)
        time = random.uniform(1, 2)
    # Calculating speed
    speed = distance / time 

    match_total = tackles + bumps + collisions
    match_intensity = ((match_total / time) 
                       * ((speed- 14) / 30)) +  (0.1 
                       * np.exp(-injury_risk) 
                       + training_load * 0.1)
    return match_intensity

def calculate_match_intensity_without_intervention(injury_risk, training_load):
    "Match intensity calculation without coaching intervention"
    tackles = random.randint(0, 8)
    bumps = random.randint(0, 5)
    collisions = random.randint(0, 30)
    distance = random.randint(11, 19)
    time = random.uniform(1, 2)
    speed = distance / time
    match_total = tackles + bumps + collisions
    match_intensity = ((match_total / time) 
                       * ((speed- 14) / 30)) +  (0.1 
                       * np.exp(-injury_risk) 
                       + training_load * 0.1)
    return match_intensity
    



def calculate_training_load(injury_risk, 
                            match_intensity, 
                            recovery_time):
    "Calculates training load for one round"
    # When a player engages with no contact drils
    if injury_risk > 0.75:
        t_total = 0
    # When a player engages with minimal contact drils
    elif 0.5 <= injury_risk <= 0.75:
        t_total = random.randint(20, 50)
        
    else:
        t_total = random.randint(50, 100)
    # A training session lasts 8 hours
    time = 8
    training_load = (t_total / 8  +  np.exp(-injury_risk) 
                     * 0.1 + match_intensity 
                     * 0.1 + recovery_time * 0.1)   
    return training_load

def calculate_training_load_without_intervention(injury_risk,
                                                 match_intensity,
                                                 recovery_time):
    "Calculates training load without coaching intervention"
    t_total = random.randint(50, 100)
    time = 8
    training_load = (t_total / 8  +  np.exp(-injury_risk) 
                     * 0.1 + match_intensity 
                     * 0.1 + recovery_time * 0.1)  
    return training_load


def calculate_recovery_time(injury_risk, match_intensity):
    "Calculates recovery time for one round"
    number_of_hours_of_sleep = random.randint(6,10)
    nutrition_level = random.randint(8, 10)
    recovery_time = (number_of_hours_of_sleep 
                     * (nutrition_level/10) + 0.1 
                     # * np.exp(-injury_risk) 
                     * injury_risk
                     + 0.1 * match_intensity)
    return recovery_time
    
def calculate_injury_risk(training_load, 
                          match_intensity, 
                          recovery_time, 
                          a, b, c):
    "Calculates injury risk for one round"
    # Adds a random value to simulate player getting an unsuspected injury
    gaussian_variation = np.random.uniform(0, 0.05)
    z = (a * training_load 
         + b * match_intensity 
         - c * recovery_time 
         + gaussian_variation)
    z = np.clip(z, -10, 10)
    total = 1 / (1 + np.exp(-(z))) 
    
    return total 

def calculate_season_injury_risk(initial_injury_risk, a, b, c):
    "Calculates injury risk for a whole season"
    # Initialises variables
    injury_risk_array = np.zeros(25)
    match_intensity_array = np.zeros(25)
    recovery_time_array = np.zeros(25)
    training_load_array = np.zeros(25)
    match_intensity_array[0] = 20
    training_load_array[0] = 30
    recovery_time_array[0] =  24
    injury_risk_array[0] = initial_injury_risk
    # Goes through a full season
    for i in range(1, 25):
        training_load_array[i] = (calculate_training_load(injury_risk_array[i - 1],  
                                                          match_intensity_array[i-1],   
                                                          recovery_time_array[i-1]))
        match_intensity_array[i] = (calculate_match_intensity(injury_risk_array[i - 1],  
                                                              training_load_array[i-1]) )
        recovery_time_array[i] = (calculate_recovery_time(injury_risk_array[i - 1], 
                                                          match_intensity_array[i-1]))
        new_risk = (calculate_injury_risk(training_load_array[i] , 
                                          match_intensity_array[i] , 
                                          recovery_time_array[i] , 
                                          a, b, c))
        injury_risk_array[i] = (0.01 * new_risk 
                                + 0.99 
                                * injury_risk_array[i-1] 
                                + np.random.uniform(-0.05, 0.05))
    return np.clip(injury_risk_array, 0, 1)


def calculate_season_injury_risk_without_intervention(initial_injury_risk, a, b, c):
    "Calculates injury risk for a whole season without intervention from coach"
    # Initialises variables
    injury_risk_array = np.zeros(25)
    match_intensity_array = np.zeros(25)
    recovery_time_array = np.zeros(25)
    training_load_array = np.zeros(25)
    match_intensity_array[0] = 20
    training_load_array[0] = 30
    recovery_time_array[0] =  24
    injury_risk_array[0] = initial_injury_risk
    # Goes through a full season
    for i in range(1, 25):
        training_load_array[i] = (calculate_training_load_without_intervention(injury_risk_array[i - 1],  
                                                          match_intensity_array[i-1],   
                                                          recovery_time_array[i-1]))
        
        match_intensity_array[i] = (calculate_match_intensity_without_intervention(injury_risk_array[i - 1],  
                                                              training_load_array[i-1]) )
        recovery_time_array[i] = (calculate_recovery_time(injury_risk_array[i - 1], 
                                                          match_intensity_array[i-1]))
        new_risk = (calculate_injury_risk(training_load_array[i] , 
                                          match_intensity_array[i] , 
                                          recovery_time_array[i] , 
                                          a, b, c))
        injury_risk_array[i] = (0.01 * new_risk 
                                + 0.99 
                                * injury_risk_array[i-1] 
                                + np.random.uniform(-0.05, 0.05))
    return np.clip(injury_risk_array, 0, 1)
```

The model above represents a dynamic, discrete, stochastic system. The dynamic nature of the model is derived from the previous values of different variables being used to calculate the current variables of each round. Random probability values introduced at the injury risk equation highlights the stochastic nature. As collisions, tackles, bumps and the number of hours slept are countable, this makes the model discrete. The model aims to calculate the injury risk over a 24-round season after being initialised with certain variables. 

This model has been simulated using python. Each of the equations that have been outlined has been written as a separate function. The functions have been designed in a way for multiple simulations to be run. If someone wanted to conduct this experiment, they are able to change the constant values in the logistic equation and are able to run a simulation of a 24-round AFL season 1000 times, as done by a for loop that constantly calls the main function. 

Injury risk has been modelled using a logistic equation because this represents the probability of a player being injured. It ensures that values that are being outputted stay between 0 and 1. It has been combined with the previous injury risk value to show how the injury risk of a player should not be calculated in isolation, but rather be calculated in consideration of previous values. The Uniform constants included in the equation represent random factors that are outside the player's control that causes an injury. These can include various muscle sprains, corked things and hand injuries. Recovery time, match intensity and training load all follow a mathematical structure of calculating a rate or time measurement and then adding extraneous variables to finalise the equation. Equations of a similar structure have been used in literature previously(Imbach et al.). 

Inside the `calculate_match_intensity()` function, tackles, bumps and collisions all have a random value associated with a specific range. Reflecting appropriate conditions of an AFL match, a player is likely to have a maximum of 8 tackles, 5 bumps and be involved in 30 collisions(“AFL Tables”). Speed is calculated by obtaining the total distance in kilometres that a player has run and dividing it by the total number of hours that a player is on the field during a match. Some AFL players have been reported to run distances up to 19 km(Bennett), which is why the maximum value is set to 19. The time that a player is on the field varies from 1 hour (which represents only a half a game and is applicable if a player has been selected as a substitute) to 2 hours, which represents a player who has not been rested on the bench.  One aspect of this function is the conditional statement, which restricts the distance and the time a player has run depending on the injury risk. This reflects the careful injury management of players. The function is meant to resemble the equation of match intensity that has been established earlier in the model. 

Inside the `calculate_training_load()` function, the number of extreme efforts is randomised but is carefully controlled depending on a player's injury. If a player is severely injured, the coaches will opt for players not to participate in activities that involve extreme efforts in order for the body to successfully recuperate. Otherwise if a player doesn't have a severe risk of injury, the player will generally participate in more extreme efforts, with an increase if a player is not severely injured. 

For the `calculate_recovery_time()` function, the number of hours of sleep is randomised between 6 to 10, while the nutrition level is randomised from a score of 8 to 10. Nutrition level is divided by 10, as this is the maximum nutrition level. 

The `calculate_injury_risk()` function uses `np.random.uniform(0, 0.05)` as a uniform variation to represent the random occurences when a player gets injured. The return statement of that function demonstrates the logistic equation that has been established in the model. 

The `calculate_season_injury_risk()` function is considered to be the main function of the program with all the above functions acting as helper functions. It first stores some initial values and then uses a for loop to simulate an entire season while calculating the overall injury risk. 

The functions `calculate_season_injury_risk_without_intervention()`, `calculate_training_load_without_intervention()`, `calculate_match_intensity_without_intervention()`, are different versions of the above functions that do not account for coaches adjusting the training load and the time a player plays in a match in order to simulate a wide range of conditions. 


# Results

**Exploration of Model**

```python
parvals = tuple(val / 10 for val in range(0, 11, 2))
x = np.arange(1, 25)
handles = []
for parval in parvals:
    y = calculate_season_injury_risk(parval, 1/200, 1/50, 1/20)[:24]
    handle = plt.plot(x, y)
    handles.append(handle[0])
    plt.scatter(x, y)

# Add labels and title
plt.xlabel("Round")
plt.ylabel("Injury Risk")
plt.title("Figure 1: Measuring Injury Risk Over an AFL Season")

# Display the plot
labels = tuple(str(parval) for parval in parvals)
plt.legend(
    handles, labels, bbox_to_anchor=(1.05, 1),
    title="Initial injury risk",
    )
plt.ylim(-0.1, 1.1)
plt.show()
```

```python
full_y = np.zeros(25)
isolated_x = []
isolated_y = []
overall_hist = []
# Running the simulation 1000 times
n_runs = 1000
# Generate input values
parvals = tuple(val / 10 for val in range(0, 11, 2))
for parval in parvals:
    isolated_x.append(parvals)
    full_y = np.zeros(25)
    individual_hist = []
    for i in range(n_runs):
        # call the main function with input parameters
        y = calculate_season_injury_risk(parval, 1/200, 1/50, 1/20)
        full_y += y
        # Get the difference between end and starting injury risk value
        individual_hist.append(y[-1] - y[0])
    overall_hist.append(individual_hist)
    isolated_y.append(float(full_y[24] / n_runs))


# Displays a table that displays the average 
dataframe = pd.DataFrame({'Initial Injury Risk Value': parvals, "Average Injury Risk at End of Season": isolated_y})
print(dataframe.to_string(index=False))

# Plot the probability density histogram
for i, hist_data in enumerate(overall_hist):
    plt.hist(hist_data, bins=30, alpha=0.5, label=f'initial_injury_risk={parvals[i]}', density=True)
plt.legend()
plt.xlabel("Difference of Injury risk from start to end of season")
plt.ylabel("Density Values")
plt.title("Figure 2: Probability Density Histogram of Final Injury Risk Difference for each Initial Injury Risk")
pass
```

# Discussion
The results support the hypothesis as it demonstrates that players with higher starting initial injury risks show a greater improvement when it comes to managing injuries compared to players with lower starting initial injury risks. In Figure 7-9, a simulation without coaching intervention was tested, which meant that the number of opportunities for a player to get injured wasn't restricted based on prior injury levels. Interestingly, in that simulation, according to Figure 9, most of the initial injury risk values that showed a decrease in overall injury risk after the season had finished, were values between 0.8 to 1.0. This was deemed to be suprising because it was expected that when coaches do not adjust their plans when players get severely injured, the injury risk is expected to increase as players would not rehabilitate properly. 

When isolating a specific trial as indicated in Figures 2, 4, 6, 8, many observable patterns can be noticed. When the initial injury risk value is 1.0, for the first 7 to 10 rounds, the values stablise around a specific value, before starting to decrease. There are also fluctuations within the results which represent random injuries that players can experience by accident and resembles the stochastic nature of the model. 

The model would be valid if the inputs such as measuring the number of physical exertions or measuring the number of tackles in a match are measured consistently and any changes that are made are smooth and gradual. This is based on how the model was designed as it relies on consistent inputs to calculate information such as match intensity and training load. Although the uniform variable provides some element of randomness, the model cannot account for abrupt changes. An example would be if a coach gets sacked and a new coach decides to implement a completely different training load. This is because the uniform variable has a specific and limited range of values which doesn't vary a lot throughout the simulation.

When calculating the match intensity, the model would be valid if conditions of the match remain similar for all 24 rounds of the season. This is because in the code for all of the 24 rounds, the number of physical contests a player experiences, only varies by the injury level of the player. However, this wouldn’t be valid when considering the match conditions over a full season. This is because in the start of the season, when players first start playing, players are more likely to engage in physical contests because they feel less fatigued and haven’t experienced a lot of match intensity. However, as the season progresses, and players start to feel more fatigued after playing a match, the number of physical contests a player starts to engage in decreases. This variation of the number of physical contests via the same level of injury risk is not represented in the model.

The model is valid assuming that players are always selected and play all 24 rounds of the match. This is because the model assumes that a player will be carefully managed throughout the season and play through some of their injuries. Playing through injury is common in the AFL as players have greater access to rehabilitation measures and medical treatments compared to amateur or semi-professional players. However, where this model fails is when it doesn’t account for fringe players. Fringe players are players who do not get selected to play for every single match, due to their performance, rather than their injury level. When players are not selected, they instead play for their state-affiliated club to get some match fitness. However, match conditions in state leagues are less intense than official AFL matches, which means there are less physical contests and a decreased likelihood of a player getting injured.

# Conclusion

AFL players play in a 24-round season and experience different types of injuries. Some injuries are caused by physical efforts that happen during training or matches, while some injuries occur via chance. The paper aimed to find out if players who had a higher initial injury risk showed an overall improvement in terms of their injury risk compared to players with a lower initial injury risk. After running many simulations, it was determined that with or without coaching intervention, players with a higher initial injury risk showed a higher decrease in injury risk at the end of the season compared to players with a lower injury risk. Future research will aim to investigate Australian Rules Footballers who play in amateur or semi-professional conditions. Modelling amateur or semi-professional players for their injury risk would be more challenging, as training standards are less consistent amongst players and rehabilitation, on average takes longer, compared to professional AFL players. Another extension of this paper would be to account for parts of the model that weren't valid. This includes accounting for players who have been dropped to the reserves after a poor performance and how this subsequently can affect injury risk.
